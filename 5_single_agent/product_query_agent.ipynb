```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain.tools import tool
from langchain.agents import create_agent

load_dotenv()
```

```python
PRODUCTS = {
    "iphone": {
        "price": 999,
        "description": "The Apple iPhone is a premium smartphone featuring a powerful processor, a high-quality camera system, a beautiful display, and long battery life. It is ideal for photography, communication, gaming, and everyday productivity.",
        "stock": 25
    },
    "macbook": {
        "price": 1999,
        "description": "The Apple MacBook is a powerful and lightweight laptop designed for professionals, students, and developers. It offers excellent performance, long battery life, a high-resolution Retina display, and is suitable for programming, AI development, and video editing.",
        "stock": 12
    },
    "headphones": {
        "price": 299,
        "description": "These premium wireless headphones deliver excellent sound quality, active noise cancellation, and a comfortable design. They are perfect for listening to music, attending online meetings, gaming, and working in noisy environments.",
        "stock": 40
    },
    "keyboard": {
        "price": 99,
        "description": "This mechanical keyboard features responsive switches, customizable RGB lighting, and an ergonomic design. It provides a comfortable typing experience and is ideal for programmers, gamers, and professionals who spend long hours working on computers.",
        "stock": 60
    }
}

llm = ChatGroq(model="openai/gpt-oss-120b", temperature = 0)

@tool
def get_product(name: str) -> str:
    """Look up a product by name and return its price, rating, stock, and description """
    p = PRODUCTS.get(name.lower())
    if not p:
        return f"Produc not found. Available:{', '.join(PRODUCTS)}"
    return str(p)

agent = create_agent(
    llm,
    tools=[get_product],
    system_prompt = "You are a helpul product assistant for an online tech store"
)
```

```python
def ask(question: str):
    result = agent.invoke({"messages":[{"role":"user", "content": question}]})
    print(result["messages"][-1].content)

ask("What is the price of macbook")
```

```python
REVIEWS = {
    "iphone": {
        "reviews_number": 1250,
        "rating": 4.8
    },
    "macbook": {
        "reviews_number": 890,
        "rating": 4.9
    },
    "headphones": {
        "reviews_number": 2100,
        "rating": 4.6
    },
    "keyboard": {
        "reviews_number": 560,
        "rating": 4.5
    }
}

@tool
def get_review(name: str) -> str:
    """Look up a product review by a product name. Return the product name, number of reviews and rating. """

    r = REVIEWS.get(name.lower())
    if not r:
        return f"Review not available for this product"
    return str(r)
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b", temperature = 0)
agent2 = create_agent(
    llm,
    tools=[get_product, get_review],
    system_prompt = "You are a helpul product assistant for an online tech store"
)

def ask2(question: str):
    result = agent2.invoke({"messages": [{"role": "user", "content": question}]})
    print(result["messages"][-1].content)
```

```python
ask2("How do people like macbook")
```
